```python
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
openai_client = OpenAI()
```

```python
def llm(prompt):
    response = openai_client.responses.create(
        model="gpt-5.6-luna",
        input=prompt,
    )
    return response.output_text
```

```python
llm("Hey, what's up?")
```
